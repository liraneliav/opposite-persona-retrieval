````python
def call_gpt(
    prompt,
    system="You are a helpful assistant.",
    max_tokens=1200,
    retries=3,
    sleep_base=3,
):
    for attempt in range(retries):
        try:
            resp = client.chat.completions.create(
                model=GPT_MODEL,
                messages=[
                    {"role": "system", "content": system},
                    {"role": "user", "content": prompt},
                ],
                max_completion_tokens=max_tokens,
            )

            msg = resp.choices[0].message
            txt = (msg.content or "").strip()
            finish_reason = resp.choices[0].finish_reason

            print("finish_reason:", finish_reason)
            print("usage:", getattr(resp, "usage", None))

            if txt:
                return txt

            print("EMPTY GPT OUTPUT, attempt:", attempt + 1)

        except Exception as e:
            print("GPT ERROR:", repr(e))

        time.sleep(sleep_base * (attempt + 1))

    return ""
````

````python
def gpt_json(prompt, system="Return valid JSON only.", max_tokens=1600):
    txt = call_gpt(
        prompt=prompt,
        system=system + "\nReturn ONLY a valid JSON object. No markdown. No explanation.",
        max_tokens=max_tokens,
    )

    if not txt:
        return {"parse_error": True, "raw": ""}

    try:
        return json.loads(txt)
    except Exception:
        cleaned = txt.replace("```json", "").replace("```", "").strip()
        start = cleaned.find("{")
        end = cleaned.rfind("}")

        if start != -1 and end != -1 and end > start:
            try:
                return json.loads(cleaned[start:end + 1])
            except Exception:
                pass

    print("JSON parse failed. Raw text:", txt[:500])
    return {"parse_error": True, "raw": txt}
````

````python
def gpt_json_with_token_fallback(prompt, system):
    # Original behavior: up to 3 attempts with 1200 tokens
    data = gpt_json(
        prompt,
        system=system,
        max_tokens=1200,
    )

    if isinstance(data, dict) and not data.get("parse_error"):
        return data

    # Only for this failed request: retry with 4000 tokens
    print("GPT failed with 1200 tokens. Retrying with 4000...")

    return gpt_json(
        prompt,
        system=system,
        max_tokens=4000,
    )
````

````python
def safe_json_loads(txt):
    txt = (txt or "").strip()
    if not txt:
        return {"parse_error": True, "raw": ""}

    try:
        return json.loads(txt)
    except Exception:
        cleaned = txt.replace("```json", "").replace("```", "").strip()
        start = cleaned.find("{")
        end = cleaned.rfind("}")
        if start != -1 and end != -1 and end > start:
            try:
                return json.loads(cleaned[start:end + 1])
            except Exception:
                pass

    return {"parse_error": True, "raw": txt}


def call_gpt_quiet(prompt, system="You are a helpful assistant.", max_tokens=1800, retries=3):
    for attempt in range(retries):
        try:
            resp = client.chat.completions.create(
                model=GPT_MODEL,
                messages=[
                    {"role": "system", "content": system},
                    {"role": "user", "content": prompt},
                ],
                max_completion_tokens=max_tokens,
            )
            txt = (resp.choices[0].message.content or "").strip()
            if txt:
                return txt
        except Exception as e:
            print("GPT error:", repr(e))

        time.sleep(2 * (attempt + 1))

    return ""


def gpt_json_quiet(prompt, system="Return valid JSON only.", max_tokens=2200):
    txt = call_gpt_quiet(
        prompt=prompt,
        system=system + "\nReturn ONLY valid JSON. No markdown. No explanation.",
        max_tokens=max_tokens,
    )
    return safe_json_loads(txt)
````

````python
# ============================================================
# Our method
# ============================================================

def runtime_pool_score(df):
    df = df.copy()

    for col in [
        "gpt_final_score",
        "keep_score",
        "persona_usefulness",
        "coverage_score",
        "uniqueness_score",
        "clarity_score",
        "redundancy_risk",
        "hnsw_similarity",
        "topic_candidate_score",
    ]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0.0)

    score = np.zeros(len(df), dtype=np.float32)

    if "gpt_final_score" in df.columns:
        score += 2.0 * df["gpt_final_score"].values
    if "keep_score" in df.columns:
        score += 1.5 * df["keep_score"].values
    if "persona_usefulness" in df.columns:
        score += 1.2 * df["persona_usefulness"].values
    if "coverage_score" in df.columns:
        score += 1.0 * df["coverage_score"].values
    if "uniqueness_score" in df.columns:
        score += 0.8 * df["uniqueness_score"].values
    if "clarity_score" in df.columns:
        score += 0.5 * df["clarity_score"].values
    if "redundancy_risk" in df.columns:
        score -= 0.8 * df["redundancy_risk"].values
    if "hnsw_similarity" in df.columns:
        score += 0.5 * df["hnsw_similarity"].values
    if "topic_candidate_score" in df.columns:
        score += 0.3 * df["topic_candidate_score"].values

    return score


def method_ours_nli_gpt(
    user_opinion,
    topic,
    expected_user_stance,
    meta,
    pool_size=60,
    k2_short=8,
):
    target_stance = opposite_stance(expected_user_stance)

    pool = meta[meta[STANCE_COL] == target_stance].copy()
    pool = pool.reset_index(drop=True)

    if pool.empty:
        return ""

    pool["_runtime_pool_score"] = runtime_pool_score(pool)
    pool = pool.sort_values("_runtime_pool_score", ascending=False).head(pool_size).copy()

    texts = pool[TEXT_COL].fillna("").astype(str).tolist()

    nli_scores = method_nli_contradiction_scores_batch(
        premise=user_opinion,
        hypotheses=texts,
        batch_size=16,
    )

    pool["nli_score"] = nli_scores
    shortlist = pool.sort_values("nli_score", ascending=False).head(k2_short).copy()

    candidates = []
    for i, row in shortlist.iterrows():
        candidates.append({
            "id": str(i),
            "text": str(row[TEXT_COL])[:700],
            "nli_score": float(row["nli_score"]),
        })

    if not candidates:
        return ""

    prompt = f"""
Topic: {topic_name(topic)}

User opinion:
{user_opinion}

Candidate opposite-stance comments:
{json.dumps(candidates, ensure_ascii=False)}

Select the single best comment for constructing an opposite-opinion persona.
Prefer direct contradiction, topic relevance, clarity, and persona usefulness.

Return a valid JSON object with this exact schema:
{{
  "best_id": "id",
  "reason": "brief reason"
}}
Do not output any text outside the JSON object.
"""

    data = gpt_json_with_token_fallback(
        prompt,
        system="You are a careful contradiction reranker. Return valid JSON only.",
    )

    best_id = str(data.get("best_id", candidates[0]["id"]))
    best = next((c for c in candidates if c["id"] == best_id), candidates[0])

    return best["text"]
````

````python
# ============================================================
# NLI models
# IMPORTANT:
# - METHOD NLI is used inside our method for NLI reranking.
# - EVALUATION NLI is used only for the reported NLI metric.
# This avoids evaluating our method with the exact same NLI model
# used inside the method.
# ============================================================

from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import numpy as np

device = "cuda" if torch.cuda.is_available() else "cpu"

METHOD_NLI_MODEL_NAME = "MoritzLaurer/multilingual-MiniLMv2-L6-mnli-xnli"
EVAL_NLI_MODEL_NAME = "MoritzLaurer/mDeBERTa-v3-base-mnli-xnli"

# Method NLI model: used inside our retrieval/reranking pipeline
method_tokenizer = AutoTokenizer.from_pretrained(METHOD_NLI_MODEL_NAME)
method_nli_model = AutoModelForSequenceClassification.from_pretrained(
    METHOD_NLI_MODEL_NAME
).to(device)
method_nli_model.eval()

# Evaluation NLI model: used only for the reported metric
eval_tokenizer = AutoTokenizer.from_pretrained(EVAL_NLI_MODEL_NAME)
eval_nli_model = AutoModelForSequenceClassification.from_pretrained(
    EVAL_NLI_MODEL_NAME
).to(device)
eval_nli_model.eval()

print("Device:", device)
print("Method NLI model:", METHOD_NLI_MODEL_NAME)
print("Evaluation NLI model:", EVAL_NLI_MODEL_NAME)
````

````python
# ============================================================
# Generic NLI contradiction helpers
# ============================================================

def find_contradiction_index(model):
    """
    Finds the contradiction label index for any HuggingFace NLI model.
    """

    id2label = model.config.id2label

    for idx, label in id2label.items():
        if "contradiction" in str(label).lower():
            return int(idx)

    raise ValueError(f"Could not find contradiction label in: {id2label}")


METHOD_CONTRA_IDX = find_contradiction_index(method_nli_model)
EVAL_CONTRA_IDX = find_contradiction_index(eval_nli_model)

print("Method contradiction index:", METHOD_CONTRA_IDX)
print("Evaluation contradiction index:", EVAL_CONTRA_IDX)


@torch.inference_mode()
def get_contradiction_score_with_model(
    premise,
    hypothesis,
    tokenizer,
    model,
    contradiction_idx,
    max_length=512,
):
    """
    Computes contradiction probability for one premise-hypothesis pair.
    """

    if not str(hypothesis).strip():
        return None

    batch = tokenizer(
        str(premise),
        str(hypothesis),
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=max_length,
    )

    batch = {k: v.to(device) for k, v in batch.items()}

    logits = model(**batch).logits
    probs = torch.softmax(logits, dim=-1)[0]

    return float(probs[contradiction_idx].detach().cpu())


@torch.inference_mode()
def get_contradiction_scores_batch_with_model(
    premise,
    hypotheses,
    tokenizer,
    model,
    contradiction_idx,
    batch_size=16,
    max_length=512,
):
    """
    Computes contradiction probabilities for one premise and many hypotheses.
    """

    scores = []

    for i in range(0, len(hypotheses), batch_size):
        batch_h = hypotheses[i:i + batch_size]
        premises = [str(premise)] * len(batch_h)

        batch = tokenizer(
            premises,
            [str(h) for h in batch_h],
            return_tensors="pt",
            truncation=True,
            padding=True,
            max_length=max_length,
        )

        batch = {k: v.to(device) for k, v in batch.items()}

        logits = model(**batch).logits
        probs = torch.softmax(logits, dim=-1)[:, contradiction_idx]

        scores.extend(probs.detach().cpu().numpy().tolist())

    return np.array(scores, dtype=np.float32)
````

````python
# ============================================================
# Method NLI functions
# Used ONLY inside our method for retrieval/reranking
# ============================================================

def method_nli_contradiction_score(premise, hypothesis):
    return get_contradiction_score_with_model(
        premise=premise,
        hypothesis=hypothesis,
        tokenizer=method_tokenizer,
        model=method_nli_model,
        contradiction_idx=METHOD_CONTRA_IDX,
    )


def method_nli_contradiction_scores_batch(premise, hypotheses, batch_size=16):
    return get_contradiction_scores_batch_with_model(
        premise=premise,
        hypotheses=hypotheses,
        tokenizer=method_tokenizer,
        model=method_nli_model,
        contradiction_idx=METHOD_CONTRA_IDX,
        batch_size=batch_size,
    )
````

````python
# ============================================================
# Evaluation NLI functions
# Used ONLY for reported NLI metric
# ============================================================

def eval_nli_contradiction_score(premise, hypothesis):
    return get_contradiction_score_with_model(
        premise=premise,
        hypothesis=hypothesis,
        tokenizer=eval_tokenizer,
        model=eval_nli_model,
        contradiction_idx=EVAL_CONTRA_IDX,
    )


def eval_nli_contradiction_scores_batch(premise, hypotheses, batch_size=16):
    return get_contradiction_scores_batch_with_model(
        premise=premise,
        hypotheses=hypotheses,
        tokenizer=eval_tokenizer,
        model=eval_nli_model,
        contradiction_idx=EVAL_CONTRA_IDX,
        batch_size=batch_size,
    )
````

````python
# Original baseline GPT stance classifier and ACU evaluation (unchanged).
def classify_persona_stance_with_gpt(output_text, topic):
    if not str(output_text).strip():
        return {
            "predicted_persona_stance": "empty",
            "stance_confidence": 0.0,
            "stance_reason": "empty output",
        }

    prompt = f"""
Topic:
{topic_name(topic)}

Persona text:
{output_text}

Classify the persona text's stance toward the topic.

Labels:
- "pro": supports the topic/movement/policy/right.
- "anti": opposes the topic/movement/policy/right.
- "unclear": unclear, mixed, irrelevant, or ambiguous.

Return valid JSON only:
{{
  "stance": "pro",
  "confidence": 0.0,
  "reason": "brief explanation"
}}
"""

    data = gpt_json_quiet(
        prompt,
        system="You are a careful stance classifier. Return valid JSON only.",
        max_tokens=800,
    )

    stance = str(data.get("stance", "unclear")).lower().strip()
    if stance not in ["pro", "anti", "unclear"]:
        stance = "unclear"

    try:
        confidence = float(data.get("confidence", 0.0))
    except Exception:
        confidence = 0.0

    confidence = max(0.0, min(1.0, confidence))

    return {
        "predicted_persona_stance": stance,
        "stance_confidence": confidence,
        "stance_reason": str(data.get("reason", ""))[:300],
    }


def compute_stance_accuracy(output_text, topic, expected_user_stance):
    target = opposite_stance(expected_user_stance)

    stance_info = classify_persona_stance_with_gpt(
        output_text=output_text,
        topic=topic,
    )

    predicted = stance_info["predicted_persona_stance"]

    return {
        **stance_info,
        "target_persona_stance": target,
        "opposite_stance_correct": bool(predicted == target),
        "opposite_stance_accuracy": 1.0 if predicted == target else 0.0,
    }


````

````python
# Freeze/verify 600-case identity and load repositories.
from collections import Counter
TEST_CASES_SHA256 = hashlib.sha256(
    json.dumps(TEST_CASES, ensure_ascii=False, separators=(',', ':')).encode('utf-8')
).hexdigest()
print('Frozen test cases SHA256:', TEST_CASES_SHA256)

def load_repository(variant, topic):
    value = REPOSITORY_FILES[variant][topic]
    if not value:
        raise ValueError(f'Missing final Step 4 repository path: {variant}/{topic}')
    path = Path(value)
    if not path.is_file():
        raise FileNotFoundError(path)
    df = pd.read_parquet(path) if path.suffix.lower()=='.parquet' else pd.read_csv(path, encoding='utf-8-sig')
    if TEXT_COL not in df.columns:
        if 'message' in df.columns: df[TEXT_COL] = df['message']
        else: raise ValueError(f'Missing {TEXT_COL} in {path}')
    if variant != 'A2':
        if STANCE_COL not in df.columns: raise ValueError(f'Missing {STANCE_COL} in {path}')
        df[STANCE_COL] = df[STANCE_COL].astype(str).str.lower().str.strip()
        df = df[df[STANCE_COL].isin(['pro','anti'])].copy()
    df = df.reset_index(drop=True)
    if len(df)==0: print('WARNING empty repository:', variant, topic)
    if variant == 'A2' and STANCE_COL in df.columns:
        print('WARNING A2 has stance_label; no filtering will be performed')
    return df, path

def file_sha256(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for chunk in iter(lambda:f.read(1024*1024),b''): h.update(chunk)
    return h.hexdigest()

REPOSITORIES = {}
REPO_HASHES = {}
for variant in RUN_VARIANTS:
    for topic in TOPIC_NAMES:
        df,path=load_repository(variant,topic)
        REPOSITORIES[(variant,topic)] = df
        REPO_HASHES[(variant,topic)] = file_sha256(path)
        print(variant,topic,'rows',len(df),'stances',df[STANCE_COL].value_counts().to_dict() if STANCE_COL in df else 'unlabeled')
````

````python
# A2 ONLY: same baseline method, except removing unavailable stance filter.
def method_ours_a2_unlabeled(
    user_opinion,
    topic,
    expected_user_stance,
    meta,
    pool_size=60,
    k2_short=8,
):
    target_stance = opposite_stance(expected_user_stance)

    pool = meta.copy()  # A2: stance filtering absent by design
    pool = pool.reset_index(drop=True)

    if pool.empty:
        return ""

    pool["_runtime_pool_score"] = runtime_pool_score(pool)
    pool = pool.sort_values("_runtime_pool_score", ascending=False).head(pool_size).copy()

    texts = pool[TEXT_COL].fillna("").astype(str).tolist()

    nli_scores = method_nli_contradiction_scores_batch(
        premise=user_opinion,
        hypotheses=texts,
        batch_size=16,
    )

    pool["nli_score"] = nli_scores
    shortlist = pool.sort_values("nli_score", ascending=False).head(k2_short).copy()

    candidates = []
    for i, row in shortlist.iterrows():
        candidates.append({
            "id": str(i),
            "text": str(row[TEXT_COL])[:700],
            "nli_score": float(row["nli_score"]),
        })

    if not candidates:
        return ""

    prompt = f"""
Topic: {topic_name(topic)}

User opinion:
{user_opinion}

Candidate opposite-stance comments:
{json.dumps(candidates, ensure_ascii=False)}

Select the single best comment for constructing an opposite-opinion persona.
Prefer direct contradiction, topic relevance, clarity, and persona usefulness.

Return a valid JSON object with this exact schema:
{{
  "best_id": "id",
  "reason": "brief reason"
}}
Do not output any text outside the JSON object.
"""

    data = gpt_json_with_token_fallback(
        prompt,
        system="You are a careful contradiction reranker. Return valid JSON only.",
    )

    best_id = str(data.get("best_id", candidates[0]["id"]))
    best = next((c for c in candidates if c["id"] == best_id), candidates[0])

    return best["text"]
````

````python
# Evaluation uses the EXACT same baseline NLI function and GPT stance classifier.
# We omit BERTScore and cosine because only NLI, ACU, and runtime were requested.
def evaluate_ablation_output(user_opinion, output_text, topic, expected_user_stance):
    valid_output = bool(str(output_text).strip())
    nli_score = eval_nli_contradiction_score(user_opinion, output_text) if valid_output else None
    stance_metrics = compute_stance_accuracy(
        output_text=output_text, topic=topic, expected_user_stance=expected_user_stance
    )
    return {
        'valid_output': valid_output,
        'eval_nli_model': EVAL_NLI_MODEL_NAME,
        'method_nli_model': METHOD_NLI_MODEL_NAME,
        'nli_contradiction': nli_score,
        **stance_metrics,
        'response_words': len(str(output_text).split()) if valid_output else 0,
    }

def run_variant_case(variant, case):
    topic=case['topic']; stance=case['expected_user_stance']; opinion=case['user_opinion']
    meta=REPOSITORIES[(variant,topic)]
    t0=time.time()
    try:
        if variant=='A2':
            output=method_ours_a2_unlabeled(opinion,topic,stance,meta,pool_size=POOL_SIZE,k2_short=K2_SHORT)
        else:
            output=method_ours_nli_gpt(opinion,topic,stance,meta,pool_size=POOL_SIZE,k2_short=K2_SHORT)
        error=''
    except Exception as e:
        output='';error=repr(e)
        print('METHOD ERROR',variant,topic,error)
    runtime_ms=int((time.time()-t0)*1000)

    metrics = evaluate_ablation_output(opinion, output, topic, stance)

    return dict(
        topic=topic,
        user_opinion=opinion,
        expected_user_stance=stance,
        method='ours_stance_nli_gpt_final_repository',
        variant=variant,
        runtime_ms=runtime_ms,
        runtime_seconds=runtime_ms / 1000,
        output_text=output,
        method_error=error,
        **metrics
    )

````

````python
# ===== RESUMABLE EXECUTION =====
# Atomic checkpoint per case. Fingerprint ensures neither the 600 cases nor input repositories change.
import os

def atomic_json(path,obj):
    path.parent.mkdir(parents=True,exist_ok=True)
    tmp=path.with_suffix(path.suffix+'.tmp')
    with open(tmp,'w',encoding='utf-8') as f:json.dump(obj,f,ensure_ascii=False)
    os.replace(tmp,path)

def chosen_cases():
    if RUN_MODE=='FULL':return list(enumerate(TEST_CASES))
    if RUN_MODE!='TEST':raise ValueError('RUN_MODE must be TEST or FULL')
    seen=Counter();picked=[]
    for i,case in enumerate(TEST_CASES):
        key=(case['topic'],case['expected_user_stance'])
        if seen[key]<TEST_CASES_PER_GROUP:
            picked.append((i,case));seen[key]+=1
    return picked

OUTPUT_ROOT.mkdir(parents=True,exist_ok=True)
for variant in RUN_VARIANTS:
    base=OUTPUT_ROOT/RUN_MODE/variant
    fingerprint=hashlib.sha256(json.dumps({
        'test_cases':TEST_CASES_SHA256,
        'repository_sha256':{topic:REPO_HASHES[(variant,topic)] for topic in TOPIC_NAMES},
        'variant':variant,'mode':RUN_MODE,'test_cases_per_group':TEST_CASES_PER_GROUP if RUN_MODE=='TEST' else None,
        'pool_size':POOL_SIZE,'k2_short':K2_SHORT,
        'method_nli_model':METHOD_NLI_MODEL_NAME,'eval_nli_model':EVAL_NLI_MODEL_NAME,
        'gpt_model':GPT_MODEL,
    },sort_keys=True).encode()).hexdigest()
    rows=[]
    for idx,case in tqdm(chosen_cases(),desc=f'Online {variant}'):
        checkpoint=base/'cases'/f'case_{idx:03d}.json'
        if checkpoint.exists():
            record=json.loads(checkpoint.read_text(encoding='utf-8'))
            if record.get('fingerprint')!=fingerprint:
                raise ValueError(f'Checkpoint mismatch; do not overwrite: {checkpoint}')
            row=record['row']
        else:
            row=run_variant_case(variant,case)
            atomic_json(checkpoint,{'fingerprint':fingerprint,'case_index':idx,'row':row})
        rows.append(row)
    result=pd.DataFrame(rows)
    result.to_csv(base/'online_results.csv',index=False,encoding='utf-8-sig')
    atomic_json(base/'manifest.json',{'fingerprint':fingerprint,'test_cases_sha256':TEST_CASES_SHA256,
        'repository_sha256':{topic:REPO_HASHES[(variant,topic)] for topic in TOPIC_NAMES},
        'n_cases':len(rows),'variant':variant,'run_mode':RUN_MODE,
        'method_nli_model':METHOD_NLI_MODEL_NAME,'eval_nli_model':EVAL_NLI_MODEL_NAME,
        'gpt_model':GPT_MODEL,'pool_size':POOL_SIZE,'k2_short':K2_SHORT})
    print(variant,'saved',len(result),'cases to',base)
````

````python
# ===== PAIRED STATISTICAL TESTS: ORIGINAL FULL vs A1-A3 =====
# Run after completing FULL and providing the original per-case baseline CSV.
# The original published aggregate (0.268 / 0.923 / 5.27) cannot be used
# for paired tests without the 600 individual baseline observations.
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon, binomtest
from statsmodels.stats.multitest import multipletests

BASELINE_RESULTS_CSV = Path(
    '/content/persona_comparison_full_results_600cases.csv'
)
BASELINE_METHOD = 'ours_stance_nli_gpt_final_repository'
CASE_COLS = ['topic', 'expected_user_stance', 'user_opinion']
N_BOOT = 5000
ALPHA = 0.05

# Map each metric to its actual column in the baseline CSV.
# Update BASELINE_COLS after inspecting print(base_raw.columns.tolist()).
BASELINE_COLS = {
    'NLI': 'nli_contradiction',
    'ACU': 'opposite_stance_accuracy',
    'Runtime': 'runtime_seconds',
}
ABLATION_COLS = {
    'NLI': 'nli_contradiction',
    'ACU': 'opposite_stance_accuracy',
    'Runtime': 'runtime_seconds',
}

if RUN_MODE != 'FULL':
    print('Statistical tests skipped: switch RUN_MODE to FULL after TEST QA.')
else:
    if not BASELINE_RESULTS_CSV.is_file():
        raise FileNotFoundError(f'Original per-case baseline CSV not found: {BASELINE_RESULTS_CSV}')
    base_raw = pd.read_csv(BASELINE_RESULTS_CSV, encoding='utf-8-sig')
    print('Baseline columns:', base_raw.columns.tolist())
    if 'method' in base_raw.columns:
        base_raw = base_raw.loc[base_raw['method'].eq(BASELINE_METHOD)].copy()
    if len(base_raw) != 600:
        raise ValueError(f'Expected 600 original baseline cases; found {len(base_raw)}. '
                         'Check BASELINE_METHOD and whether the file is per-case.')
    required = CASE_COLS + list(BASELINE_COLS.values())
    missing = [c for c in required if c not in base_raw.columns]
    if missing:
        raise ValueError(f'Baseline columns missing: {missing}. '
                         'Adjust BASELINE_COLS to match the printed column names.')
    if base_raw.duplicated(CASE_COLS).any():
        raise ValueError('Duplicate baseline test-case keys.')
    baseline = base_raw[required].copy()
    baseline = baseline.rename(columns={v: k for k, v in BASELINE_COLS.items()})
    if set(baseline['topic']) != {'blm','guns','samesex'}:
        raise ValueError('Baseline topic names differ; normalize only after verifying case identity.')

    def bootstrap_mean_ci(diff, seed=42):
        d = np.asarray(diff, dtype=float)
        rng = np.random.default_rng(seed)
        estimates = [rng.choice(d, size=len(d), replace=True).mean()
                     for _ in range(N_BOOT)]
        return np.quantile(estimates, [0.025, 0.975])

    def rank_biserial(diff):
        d = np.asarray(diff, dtype=float)
        d = d[d != 0]
        if not len(d): return 0.0
        ranks = pd.Series(np.abs(d)).rank(method='average').to_numpy()
        return float((ranks[d > 0].sum() - ranks[d < 0].sum()) / ranks.sum())

    stats_rows = []
    for variant in RUN_VARIANTS:
        ablation_path = OUTPUT_ROOT / RUN_MODE / variant / 'online_results.csv'
        ablation = pd.read_csv(ablation_path, encoding='utf-8-sig')
        if len(ablation) != 600 or ablation.duplicated(CASE_COLS).any():
            raise ValueError(f'{variant}: expected 600 unique cases.')
        for metric, column in ABLATION_COLS.items():
            if column not in ablation:
                raise ValueError(f'{variant}: missing {column}')
        selected = ablation[CASE_COLS + list(ABLATION_COLS.values())].copy()
        selected = selected.rename(columns={v: k for k, v in ABLATION_COLS.items()})
        paired = baseline.merge(selected, on=CASE_COLS, how='inner',
                                suffixes=('_full', '_ablation'), validate='one_to_one')
        if len(paired) != 600:
            raise ValueError(f'{variant}: only {len(paired)} matched baseline cases.')
        for metric in ['NLI','ACU','Runtime']:
            x = pd.to_numeric(paired[f'{metric}_full'], errors='coerce')
            y = pd.to_numeric(paired[f'{metric}_ablation'], errors='coerce')
            valid = x.notna() & y.notna()
            if not valid.any():
                raise ValueError(f'{variant}/{metric}: no valid paired observations.')
            full = x[valid].to_numpy(dtype=float)
            abl = y[valid].to_numpy(dtype=float)
            d = full - abl
            ci = bootstrap_mean_ci(d)
            row = dict(variant=variant, metric=metric, n_pairs=len(d),
                       full_mean=full.mean(), ablation_mean=abl.mean(),
                       mean_difference_full_minus_ablation=d.mean(),
                       bootstrap_95CI_low=ci[0], bootstrap_95CI_high=ci[1])
            if metric == 'ACU':
                if not (np.isin(full, [0,1]).all() and np.isin(abl, [0,1]).all()):
                    raise ValueError('ACU must be binary 0/1 for exact McNemar.')
                b = int(((full == 1) & (abl == 0)).sum())
                c = int(((full == 0) & (abl == 1)).sum())
                p = float(binomtest(min(b,c), n=b+c, p=0.5).pvalue) if b+c else 1.0
                row.update(test='Exact McNemar', statistic=np.nan, p_raw=p,
                           rank_biserial=np.nan, discordant_full_only=b,
                           discordant_ablation_only=c)
            else:
                if np.all(d == 0):
                    w, p = 0.0, 1.0
                else:
                    result = wilcoxon(full, abl, alternative='two-sided',
                                      zero_method='wilcox', method='auto')
                    w, p = float(result.statistic), float(result.pvalue)
                row.update(test='Wilcoxon signed-rank', statistic=w, p_raw=p,
                           rank_biserial=rank_biserial(d),
                           discordant_full_only=np.nan,
                           discordant_ablation_only=np.nan)
            stats_rows.append(row)

    stats = pd.DataFrame(stats_rows)
    stats['p_holm'] = np.nan
    stats['significant_after_holm'] = False
    for metric, ix in stats.groupby('metric').groups.items():
        ix = list(ix)
        rejected, corrected, _, _ = multipletests(
            stats.loc[ix, 'p_raw'].to_numpy(), alpha=ALPHA, method='holm')
        stats.loc[ix, 'p_holm'] = corrected
        stats.loc[ix, 'significant_after_holm'] = rejected
    out = OUTPUT_ROOT / RUN_MODE / 'statistical_tests'
    out.mkdir(parents=True, exist_ok=True)
    output_path = out / 'paired_full_vs_ablations.csv'
    stats.to_csv(output_path, index=False, encoding='utf-8-sig')
    display(stats.round(6))
    print('Saved:', output_path)
    print('NLI uses only matched valid outputs; check valid_output_rate separately.')
    print('A positive Full-minus-ablation difference favors Full for NLI/ACU;')
    print('a negative difference favors Full for Runtime.')
````
